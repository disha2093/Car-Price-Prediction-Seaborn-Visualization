```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# display setting
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

df=pd.read_csv('/content/CarPrice_Assignment.csv')

display(df)
df.info()


```

```python
print('\n check null values')
print(df.isnull().sum())

plt.boxplot(df['price'])
plt.title('Price Distubution')
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

```python
print("\n Find the highest selling price.")

Highest_selling_price=df['price'].sort_values(ascending=False)
print(Highest_selling_price.head(1))


```

```python
print("\n Find the lowest selling price.")
lowest_selling_price=df['price'].sort_values(ascending=True)
print(lowest_selling_price.head(1))
```

```python
print('\nFind the average selling price.')
Avg_selling_price= df['price'].mean()
print(Avg_selling_price)
```

```python
print('\n Which fuel type has the highest average selling price?')


df.groupby('fueltype')['price'].mean().head(1)
```

```python
print('\n Which transmission type is most common?')
# Calculate the counts of each drivewheel type for the pie chart
transmission_counts = df['drivewheel'].value_counts()
print(transmission_counts)

# Create an explode list matching the number of categories
# Explode the most common category (transmission_counts.index[0]) slightly
explode_list = [0.1 if label == transmission_counts.index[0] else 0 for label in transmission_counts.index]

plt.figure(figsize=(8, 8))
plt.pie(
    transmission_counts, # Pass the numerical counts to the pie chart
    labels=transmission_counts.index, # Use the drivewheel types as labels
    autopct='%1.1f%%', # Display percentages on the slices
    shadow=True,
    explode=explode_list, # Use the dynamically created explode list
    startangle=90
)
plt.title('Transmission Type Share') # Update title to reflect the content
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()
plt.show()
```

```python

print('\nDisplay all diesel cars.')

diesel_cars = df.loc[
    df['fueltype'] == 'diesel',
    ['CarName', 'fueltype']
]

print(diesel_cars)


```

```python
print('\nCount cars by fuel type.')

count=df.groupby('fueltype')['CarName'].value_counts()


print(count)
```

```python
print('\nAverage selling price by fuel type.')

count=df.groupby('fueltype')['price'].mean()


print(count)

```

```python
print('\nCreate a Price_Category column (Budget/Premium/Luxury)')

df['Price_Category'] = pd.cut(
    df['price'],
    bins=3,
    labels=['Budget', 'Premium', 'Luxury']
)

print(df[['CarName', 'price', 'Price_Category']].head())





```

```python
print('Generate a final dashboard summary with:')
print("=" * 60)
print("Car Analytics Dashboard")
print("=" * 60)
print('Total cars:', len(df))
print('Average selling price:', df['price'].mean())
print('Highest selling price:', df['price'].max())
print('Lowest selling price:', df['price'].min())
print('Most common fuel type:', df['fueltype'].mode().iloc[0])
print('Top 5 expensive cars:')
display(df.nlargest(5, 'price')[['CarName', 'price']])
```

# Seaborn Data Visualization

The following cells extend the car-price analysis with Seaborn visualizations.
They cover price distribution, categorical comparisons, relationships between
engine/car attributes and price, correlation, and price categories.

```python
# Import Seaborn for visualization
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", context="notebook")
```

```python
# 1. Distribution of car prices
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='price', kde=True, bins=30)
plt.title('Distribution of Car Prices')
plt.xlabel('Price')
plt.ylabel('Number of Cars')
plt.tight_layout()
plt.show()
```

```python
# 2. Average selling price by fuel type
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='fueltype', y='price', estimator='mean', errorbar=None)
plt.title('Average Selling Price by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Price')
plt.tight_layout()
plt.show()
```

```python
# 3. Price comparison by body style
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='carbody', y='price')
plt.title('Car Price Distribution by Body Style')
plt.xlabel('Body Style')
plt.ylabel('Price')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# 4. Price comparison by drive wheel
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='drivewheel', y='price')
plt.title('Car Price Distribution by Drive Wheel')
plt.xlabel('Drive Wheel')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
# 5. Horsepower vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='horsepower', y='price', hue='fueltype', style='fueltype', s=80)
plt.title('Horsepower vs Car Price')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
# 6. Engine size vs price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='enginesize', y='price', hue='carbody', s=75)
plt.title('Engine Size vs Car Price')
plt.xlabel('Engine Size')
plt.ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
# 7. Number of cars by fuel type
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='fueltype')
plt.title('Number of Cars by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
plt.tight_layout()
plt.show()
```

```python
# 8. Number of cars by body style
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='carbody', order=df['carbody'].value_counts().index)
plt.title('Number of Cars by Body Style')
plt.xlabel('Body Style')
plt.ylabel('Number of Cars')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# 9. Price category distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Price_Category', order=['Budget', 'Premium', 'Luxury'])
plt.title('Number of Cars in Each Price Category')
plt.xlabel('Price Category')
plt.ylabel('Number of Cars')
plt.tight_layout()
plt.show()
```

```python
# 10. Correlation heatmap for numerical variables
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14, 10))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5
)
plt.title('Correlation Heatmap of Numerical Car Features')
plt.tight_layout()
plt.show()
```

```python
# 11. Top 10 most expensive cars
top10 = df.nlargest(10, 'price').sort_values('price', ascending=True)

plt.figure(figsize=(10, 6))
sns.barplot(data=top10, x='price', y='CarName', errorbar=None)
plt.title('Top 10 Most Expensive Cars')
plt.xlabel('Price')
plt.ylabel('Car Name')
plt.tight_layout()
plt.show()
```

```python
# 12. Pairwise relationship of key numerical features
features = ['price', 'horsepower', 'enginesize', 'curbweight']
sns.pairplot(df[features], diag_kind='hist')
plt.suptitle('Relationships Among Price and Key Numerical Features', y=1.02)
plt.show()
```
